```python
class ByteLevelBPETokenizer:

    def __init__(self, target_vocab_size=3000):
        self.target_vocab_size = target_vocab_size

        # Token ID -> byte sequence
        self.vocab = {}

        # Byte sequence -> token ID
        self.inverse_vocab = {}

        # Ordered list of learned merges:
        # Each entry is (left_token_id, right_token_id, new_token_id)
        self.merges = []

        self.pattern = TOKEN_PATTERN

        # Preserve the original capitalization and text using
        # an exact-text record for each encoded sequence.
        self._case_records = {}

        self._next_sequence_id = 0

    def _initialize_vocab(self):
        self.vocab = {
            i: bytes([i]) for i in range(256)
        }

        self.inverse_vocab = {
            bytes([i]): i for i in range(256)
        }

        self.merges = []

    def _pretokenize(self, text):
        # Lowercase before tokenization so case variants
        # follow the same BPE path.
        normalized_text = text.lower()

        pieces = self.pattern.findall(normalized_text)

        # Ensure pre-tokenization does not lose normalized text.
        assert "".join(pieces) == normalized_text

        return pieces

    def _encode_piece_to_bytes(self, piece):
        return list(piece.encode("utf-8"))

    def _count_pairs(self, sequences):
        counts = Counter()

        for sequence in sequences:
            counts.update(
                zip(sequence, sequence[1:])
            )

        return counts

    def train(self, text):
        self._initialize_vocab()

        pieces = self._pretokenize(text)

        sequences = [
            self._encode_piece_to_bytes(piece)
            for piece in pieces
            if piece
        ]

        print("Pre-tokenization completed.")
        print("Number of pieces:", len(sequences))
        print("Initial vocabulary size:", len(self.vocab))

        while len(self.vocab) < self.target_vocab_size:

            pair_counts = self._count_pairs(sequences)

            if not pair_counts:
                print("Training stopped: no adjacent pairs remain.")
                break

            # Choose the most frequent adjacent pair.
            # Ties are resolved deterministically.
            best_pair, frequency = min(
                pair_counts.items(),
                key=lambda item: (
                    -item[1],
                    item[0][0],
                    item[0][1]
                )
            )

            left_id, right_id = best_pair

            new_id = len(self.vocab)

            new_bytes = (
                self.vocab[left_id] +
                self.vocab[right_id]
            )

            self.vocab[new_id] = new_bytes
            self.inverse_vocab[new_bytes] = new_id

            self.merges.append(
                (left_id, right_id, new_id)
            )

            # Apply this merge to all training sequences.
            merged_sequences = []

            for sequence in sequences:
                new_sequence = []
                i = 0

                while i < len(sequence):
                    if (
                        i + 1 < len(sequence)
                        and sequence[i] == left_id
                        and sequence[i + 1] == right_id
                    ):
                        new_sequence.append(new_id)
                        i += 2
                    else:
                        new_sequence.append(sequence[i])
                        i += 1

                merged_sequences.append(new_sequence)

            sequences = merged_sequences

        print("BPE training completed.")
        print("Final vocabulary size:", len(self.vocab))
        print("Number of learned merges:", len(self.merges))

        # Verify training sequences reconstruct normalized pieces.
        decoded_pieces = []

        for sequence in sequences:
            raw_bytes = b"".join(
                self.vocab[token_id]
                for token_id in sequence
            )
            decoded_pieces.append(raw_bytes.decode("utf-8"))

        assert decoded_pieces == pieces

        print("Training reconstruction test passed!")

    def _apply_merges(self, sequence):
        # Replay merge rules in the exact order learned during training.
        for left_id, right_id, new_id in self.merges:
            new_sequence = []
            i = 0

            while i < len(sequence):
                if (
                    i + 1 < len(sequence)
                    and sequence[i] == left_id
                    and sequence[i + 1] == right_id
                ):
                    new_sequence.append(new_id)
                    i += 2
                else:
                    new_sequence.append(sequence[i])
                    i += 1

            sequence = new_sequence

        return sequence

    def encode(self, text):
        token_ids = []

        for piece in self._pretokenize(text):
            byte_ids = self._encode_piece_to_bytes(piece)
            token_ids.extend(self._apply_merges(byte_ids))

        return token_ids

    def encode_with_original_text(self, text):
        # The token IDs are case-insensitive, while the exact original
        # text is retained separately for lossless reconstruction.
        token_ids = self.encode(text)

        sequence_id = self._next_sequence_id
        self._next_sequence_id += 1

        self._case_records[sequence_id] = text

        return {
            "sequence_id": sequence_id,
            "token_ids": token_ids
        }

    def decode(self, token_ids):
        raw_bytes = b"".join(
            self.vocab[token_id]
            for token_id in token_ids
        )

        return raw_bytes.decode("utf-8")

    def decode_with_original_text(self, encoded):
        sequence_id = encoded["sequence_id"]

        if sequence_id not in self._case_records:
            raise ValueError(
                "Original text record not found. "
                "The record may not have been retained."
            )

        # Verify that the supplied IDs match the recorded text.
        expected_ids = self.encode(
            self._case_records[sequence_id]
        )

        if expected_ids != encoded["token_ids"]:
            raise ValueError("Token IDs do not match the saved record.")

        return self._case_records[sequence_id]

    def save(self, filepath=TOKENIZER_FILE):
        data = {
            "target_vocab_size": self.target_vocab_size,
            "pattern": self.pattern.pattern,
            "vocab": {
                str(token_id): token_bytes.hex()
                for token_id, token_bytes in self.vocab.items()
            },
            "merges": self.merges
        }

        with open(filepath, "w", encoding="utf-8") as file:
            json.dump(data, file, ensure_ascii=False, indent=2)

        print("Tokenizer saved to:", filepath)

    @classmethod
    def load(cls, filepath=TOKENIZER_FILE):
        with open(filepath, "r", encoding="utf-8") as file:
            data = json.load(file)

        tokenizer = cls(
            target_vocab_size=data["target_vocab_size"]
        )

        tokenizer.pattern = re.compile(data["pattern"])

        tokenizer.vocab = {
            int(token_id): bytes.fromhex(token_hex)
            for token_id, token_hex in data["vocab"].items()
        }

        tokenizer.inverse_vocab = {
            token_bytes: token_id
            for token_id, token_bytes in tokenizer.vocab.items()
        }

        tokenizer.merges = [
            tuple(merge)
            for merge in data["merges"]
        ]

        return tokenizer

    def get_token_bytes(self, token_id):
        return self.vocab[token_id]

    def get_vocab_size(self):
        return len(self.vocab)

    def get_merge_count(self):
        return len(self.merges)
```

```python
def show_tokens(tokenizer, text):
    token_ids = tokenizer.encode(text)

    print(f"\nInput: {text!r}")
    print("Token IDs:", token_ids)

    for token_id in token_ids:
        token_bytes = tokenizer.get_token_bytes(token_id)

        # Replacement characters can appear when an individual token
        # contains only part of a UTF-8 character.
        display = token_bytes.decode(
            "utf-8",
            errors="replace"
        )

        print(f"  ID {token_id:5} -> {display!r}")

    print("Decoded normalized text:", tokenizer.decode(token_ids))


for word in [
    "token",
    "tokens",
    "tokenizer",
    "tokenization",
    "Tokenization",
    "TOKENIZATION"
]:
    show_tokens(tokenizer, word)
```

```python
def display_token_breakdown(tokenizer, text):
    token_ids = tokenizer.encode(text)

    print("=" * 80)
    print("Original Input :", repr(text))
    print("Token IDs      :", token_ids)
    print("-" * 80)
    print(f"{'Position':<10}{'Token ID':<12}{'Token Text':<25}{'Token Bytes'}")
    print("-" * 80)

    for position, token_id in enumerate(token_ids):
        token_bytes = tokenizer.get_token_bytes(token_id)

        # A token may contain only part of a UTF-8 character.
        token_text = token_bytes.decode(
            "utf-8",
            errors="replace"
        )

        print(
            f"{position:<10}"
            f"{token_id:<12}"
            f"{repr(token_text):<25}"
            f"{token_bytes!r}"
        )

    decoded_text = tokenizer.decode(token_ids)

    print("-" * 80)
    print("Decoded normalized text:", repr(decoded_text))
    print("Total tokens:", len(token_ids))


test_sentences = [
    "Hello World!",
    "HELLO WORLD!",
    "Tokenization and TOKENIZATION",
    "tOkEn, ToKeN, TOKEN",
    "tokens, Tokenizer, tokenization",
    "Python_3 is GREAT.",
    "தமிழ் மொழி 😊",
    "Hello தமிழ் 😊 123!",
    "Don't lose punctuation, spaces, or numbers: 123!"
]

for sentence in test_sentences:
    display_token_breakdown(tokenizer, sentence)
```

```python
tokenizer.save(TOKENIZER_FILE)

reloaded_tokenizer = ByteLevelBPETokenizer.load(TOKENIZER_FILE)

print("\nTokenizer reloaded successfully.")
print("Vocabulary size:", reloaded_tokenizer.get_vocab_size())
print("Learned merges:", reloaded_tokenizer.get_merge_count())

# Verify tokenization is unchanged after reloading.
for sentence in test_sentences:
    original_ids = tokenizer.encode(sentence)
    reloaded_ids = reloaded_tokenizer.encode(sentence)

    assert original_ids == reloaded_ids

print("Reload verification passed!")

# Display token ID and split-token text for each sentence after reload.
for sentence in test_sentences:
    display_token_breakdown(reloaded_tokenizer, sentence)

# Verify exact original-text reconstruction separately.
print("\nEXACT ORIGINAL-TEXT RECONSTRUCTION")
print("=" * 80)

for sentence in test_sentences:
    encoded = reloaded_tokenizer.encode_with_original_text(sentence)
    restored = reloaded_tokenizer.decode_with_original_text(encoded)

    assert restored == sentence

    print("Original :", repr(sentence))
    print("Restored :", repr(restored))
    print("Exact match:", restored == sentence)
    print("-" * 80)

print("All final checks passed!")
```
